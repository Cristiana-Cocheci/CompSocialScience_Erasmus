```python
def language_similarity(node1, node2):
    language_family_1 = node1['language_family']
    language_family_2 = node2['language_family']

    # split string into a set of language families
    set_1 = set([int(x) for x in language_family_1.split(",")])
    set_2 = set([int(x) for x in language_family_2.split(",")])

    intersection = set.intersection(set_1, set_2)
    reunion = set_1.union(set_2)
    if len(intersection) == 0:
        # No language family in common
        return 0 
    elif len(intersection) == len(reunion):
        # All language families in common
        return 2
    else:
        # Some language families in common
        return 1
```
