```python
ka2_df_cleaned = ka2_df[['Mobility Year', 'Mobility StartMonth', 'Activity (mob)', 'Special Needs', 'Fewer Opportunities', 'Disadvantaged Background', 'Sending Country', 'Sending Organization', 'Receiving Country', 'Receiving Organization', 'Actual Participants (Contracted Projects)']].copy()
ka2_df_cleaned = ka2_df_cleaned.replace('-', 'unknown')
ka2_df_cleaned = ka2_df_cleaned.replace('', 'unknown')

ka2_df_cleaned.head()
```

```python
ka2_df_cleaned.to_csv('../data/ka2_mobility_data_cleaned.csv', index=False)
```

```python
unique_orgs = pd.concat([ka2_df_cleaned['Sending Organization'], ka2_df_cleaned['Receiving Organization']]).unique()
print(f"Number of unique organizations: {len(unique_orgs)}")
unique_lower_orgs = [org.lower() for org in unique_orgs]
print(f"Number of unique organizations (case-insensitive): {len(set(unique_lower_orgs))}")
```

```python
ka2_df_cleaned['Sending Organization'] = ka2_df_cleaned['Sending Organization'].str.lower()
ka2_df_cleaned['Receiving Organization'] = ka2_df_cleaned['Receiving Organization'].str.lower()
ka2_df_cleaned.to_csv('../data/ka2_mobility_data_cleaned.csv', index=False)
```

```python
unique_sending_countries = ka2_df['Sending Country'].unique()
unique_receiving_countries = ka2_df['Receiving Country'].unique()
print(f"Unique Sending Countries: {len(unique_sending_countries)}")
print(unique_sending_countries)
print(f"\nUnique Receiving Countries: {len(unique_receiving_countries)}")
print(unique_receiving_countries)

unique_countries = set(unique_sending_countries) | set(unique_receiving_countries)
print(f"\nUnique Countries (Sending and Receiving): {len(unique_countries)}")
print(unique_countries)
```
